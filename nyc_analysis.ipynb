```python
X=df_clean.drop('room_type',axis=1)
y=df_clean['room_type']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42,stratify=y)
```

```python
##build numeric and categorical pipeline
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder,PowerTransformer
numeric_pipeline=Pipeline([
    ('imputer',SimpleImputer(strategy='median')),

    ('scaler',StandardScaler())
])
categorical_pipeline=Pipeline([
    ('imputer',SimpleImputer(strategy='most_frequent')),
    ('encoder',OneHotEncoder(handle_unknown='ignore'))
])
```

```python
from sklearn.compose import ColumnTransformer
preprocessor=ColumnTransformer([
    ('numeric_trf',numeric_pipeline,num_cols),
    ('categoric_trf',categorical_pipeline,cat_cols)
])
```

```python
preprocessor
```

```python
##build the model pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier
from sklearn.model_selection import cross_val_score
models={
    'Logistic Regression':LogisticRegression(class_weight="balanced",random_state=42),
    'Decision Tree':DecisionTreeClassifier(class_weight="balanced",random_state=42),
    "RandomForest":RandomForestClassifier(class_weight="balanced",random_state=42),
    "GradientBoosting":GradientBoostingClassifier(random_state=42)
}
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
# Fix the column lists: remove target from features and fix trailing spaces
num_cols = [c.strip() for c in num_cols]
cat_cols = [c for c in cat_cols if c != 'room_type']

# Update X_train/X_test column names to ensure no trailing spaces
X_train.columns = [c.strip() for c in X_train.columns]
X_test.columns = [c.strip() for c in X_test.columns]

# Redefine preprocessor with correct columns
preprocessor = ColumnTransformer([
    ('numeric_trf', numeric_pipeline, num_cols),
    ('categoric_trf', categorical_pipeline, cat_cols)
])

# Run evaluation
for name, model in models.items():
    pipe = Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])
    acc = cross_val_score(pipe, X_train, y_train, cv=5, scoring='accuracy').mean()
    f1 = cross_val_score(pipe, X_train, y_train, cv=5, scoring='f1_macro').mean()
    print(f"{name} -> Accuracy: {acc:.4f}, F1-Macro: {f1:.4f}")
```

```python
from sklearn.model_selection import RandomizedSearchCV
best_pipeline=Pipeline([
    ('preprocessor',preprocessor),
    ('classifier',RandomForestClassifier(class_weight='balanced',random_state=42))
])
param_distribution={
    "classifier__n_estimators":[100,150,200,300],
    "classifier__max_depth":[8,12,15,20,None],
    "classifier__min_samples_split":[2,5,10]
}
search=RandomizedSearchCV(best_pipeline,param_distribution,cv=2,scoring='accuracy',n_iter=5,random_state=42)
```

```python
search.fit(X_train,y_train)
```

```python
search.best_params_
```

```python
best_model_tuned=search.best_estimator_
```

```python
search.best_score_
```

```python
y_pred=best_model_tuned.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score,f1_score
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(f1_score(y_test,y_pred,average='micro'))
```
